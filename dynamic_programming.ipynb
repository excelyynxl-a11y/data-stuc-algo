```python
N = 6
memo = [-1] * (N + 1)
memo[0] = 0
memo[1] = 1 

def fib_dpa_top_down(n):
    '''
    Top-down approach.
    Recursive.
    Time-comp = O(n)
    '''
    if memo[n] != -1:
        return memo[n] # computed before 

    memo[n] = fib_dpa_top_down(n-1) + fib_dpa_top_down(n-2)
    return memo(n)

print(fib_dpa_top_down(N))
```

```python
N = 6
memo = [-1] * (N + 1)
memo[0] = 0
memo[1] = 1 

def fib_dpa_bottom_up(n):
    '''
    Top-bottom approach.
    Iterative 
    Time-comp = O(n)
    '''
    for i in range(2, n):
        memo[i] = memo[i-1] + memo[i-2]
```

```python
def fib_brute_force(n):
    '''
    Brute force using reccursion.
    Time-comp = o(2^n)
    '''
    if n == 0:
        return 0
    elif n == 1:
        return 2 
    return fib_brute_force(n-1) + fib_brute_force(n-2)
```

```python
def coin_change_bottom_up(sample_space, target_sum):
    '''
    Time-comp = O(NM), N = target_sum, M = size of sample_space .
    Bottom-up approach. 
    Iterative.
    '''
    n = target_sum
    m = len(sample_space)
    memo = [-1] * (n + 1) # initialise memo 
    memo[0] = 0 

    for value in range(1, n):
        for j in range(m): # loop through element in sample_space to be obtain balance
            if sample_space[j] <= value: # only compute balance if the sample_space element < value of target_sum
                balance = value - sample_space[j]
                count = 1 + memo[balance]
                if count < memo[value]: # found a new optimal (smaller number of coins required to sum up tp target_sum
                    memo[value] = count 

    return memo[n]
```

```python
def knapsack(items, weight_limit):
    ''''
    item = {weight, profit}
    Time-comp = O(NM), N = target_sum, M = size of sample_space .
    Bottom-up approach. 
    Iterative.
    '''
    n = weight_limit 
    m = len(items)
    memo  = [0] * n 

    for weight in range(1, n+1):
        for j in range(m): # loop through each items to get balance
            if items[j].weight <= weight: # only compute balance if item.weight <= weight
                balance = weight - items[j].weight 
                profit = items[j].profit + memo[balance]
                if profit > memo[weight]: # found new optimal profit
                    memo[weight] = profit 

    return memo[weight_limit]

```
