```python

def bla(n):
    # base case 
    if n <= 3:                  # O(1)
        return 1                # O(1)
    # general case
    else:
        for i in range(n):      # O(n)
            bla(n - 1)          # O(T(n-1))

        return bla(n - 1) + bla(n - 1) # O(2*T(n-1))

# T(n) = a                        , if n <= 3
# T(n) = n*T(n-1) + 2*T(n-1) + c  , if n > 3
```

```python

def bla(n):
    # base case 
    if n <= 3: 
        for i in range(n):      # O(1)
            print('HELLO')      # O(1)
        return 1                # O(1)
    # general case
    else:
        for i in range(n):      # O(n)
            bla(n / 2)          # O(T(n-1))

        return bla(n - 1) + bla(n - 1) # O(2*T(n-1))

# T(n) = a                        , if n <= 3
# T(n) = n*T(n/2) + 2*T(n-1) + c  , if n > 3
```
