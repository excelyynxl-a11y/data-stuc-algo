## Proving Invariant 

# To prove a loop invariant:
- initialization:
- maintainance:
- termination: 

```python

my_list = []
my_sum = 0
n = len(my_list)
i = 1
# initialization:
# when: i = 1
# then: my_sum = 0
# because: my_sum of even numbers in a my_list[1..0] which is an empty list.
#          The sum of all even numbers in an empty list is 0

while i <= n:
    # loop invariant (START): my_sum is the sum of even number sin my_list[1...i-1]

    if my_list[i] % 2 == 0:
        my_sum += my_list[i]

    # maintainance:
    # assume that my_sum is sum of even numbers in my_list[1...i-1]
    # if my_list[i] is even, my_sum already have sum of even numbers in my_list[1...i-1] and since my_list[i] is even,
    # we add it to my_sum, so now my_sum is the sum of numbers in my_list[1...i]
    # if my_list[i] is odd, my_sum already have sum of even numbers in my_list[1...i-1] and since my_list[i] is odd,
    # we don't add it to my_sum, so now my_sum is the sum of numbers in my_list[1...i]

    # loop invariant (END): my_sum is the sum of even numbers in my_list[1...i]

    i = i + 1 

# termination:
# when: i = n+1
# then: my_sum is correct
# because: my_sum is the sum of even numbers in my_list[1...n] which is my entire list.
```

```python
def COUNT(A[1..n], target):
    """
    Loop invariant: At start of iteration-i, count is the number
    of occurance of target in list A[1...i-1], 
    where we consider A[1...0] to be an empty list.  
    """
    count = 0
    i =1

    # initialisation: At the start of the first iteration,
    # when: i = 1
    # then: count = 0.
    # because: A[1...i-1] = A[1...0] which is an empty list and the 
    #          number of occurance of target in an empty list is 0.

    while i <= n:
        # loop invariant (START): count is the number or occurance of target in A[1...i-1]

        if A[i] == target:
            count = count +1

        # maintainance:
        # Assume that invariant holds at the start of k-iteration.
        # So count is the number of occurance of target in A[1...i-1].
        # During the k-iteration of the loop, if A[k] == target, count is incremented, which count is number of occurance of target in A[1...k]
        # Else, if A[k] != target, we dont change count, which count is number of occurance of target in A[1...k]. 
        # Since we know the invariant holds at the start, by induction it holds for all values of i, including when i = n + 1, so the invariant holds.

        # loop invariant (END): count is the number or occurance of target in A[1...i]
        i =i+1

    # termination: 
    # when: i = n + 1, it no longer satisfy "while i <= n" and exit the loop
    # then: count is the number of occurance of target in entire A
    # because: A[1...i-1] = A[1...n], which is the entire list.
    
    # Hence the algorithm is correct.
    return count
```
